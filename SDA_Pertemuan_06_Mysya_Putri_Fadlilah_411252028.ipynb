### 1. Membuat fungsi countdown(n) dan menjalankannya untuk n = 5

```python
from __future__ import annotations

from typing import Any


def countdown(n: int) -> None:
    if n == 0:
        print("Selesai")
        return
    print(n)
    countdown(n - 1)


countdown(5)
```

### 2. Menampilkan "Masuk n" sebelum pemanggilan rekursif dan "Keluar n" setelahnya

```python
def countdown(n: int) -> None:
    if n == 0:
        print("Selesai (base case, n = 0)")
        return
    print(f"Masuk {n}")
    countdown(n - 1)
    print(f"Keluar {n}")


countdown(5)
```

Pesan "Masuk" tercetak saat pemanggilan bergerak turun (5 sampai 1), sedangkan pesan "Keluar" tercetak saat pemanggilan kembali naik (1 sampai 5). Pemanggilan yang paling akhir masuk (n = 1) selesai paling dulu, sesuai cara kerja *call stack* (LIFO).

### 3. Membuat fungsi recursive_sum_linear(data, index=0) tanpa slicing

```python
def recursive_sum_linear(data: list[int], index: int = 0) -> int:
    if index == len(data):
        return 0
    return data[index] + recursive_sum_linear(data, index + 1)
```

### 4. Menguji list kosong, satu elemen, dan lima elemen

```python
print("List kosong :", recursive_sum_linear([]))
print("Satu elemen :", recursive_sum_linear([7]))
print("Lima elemen :", recursive_sum_linear([10, 20, 30, 40, 50]))
```

| Input | Hasil yang diharapkan | Hasil program |
|---|---|---|
| `[]` | 0 | 0 |
| `[7]` | 7 | 7 |
| `[10, 20, 30, 40, 50]` | 150 | 150 |

### 5. Dry run untuk input `[4, 6, 2]`

Panjang list = 3, jadi base case tercapai saat `index == 3`.

| Pemanggilan | index | index == 3? | Yang dihitung | Nilai kembali |
|---|---|---|---|---|
| 1 | 0 | Tidak | 4 + recursive_sum_linear(data, 1) | 12 (4 + 8) |
| 2 | 1 | Tidak | 6 + recursive_sum_linear(data, 2) | 8 (6 + 2) |
| 3 | 2 | Tidak | 2 + recursive_sum_linear(data, 3) | 2 (2 + 0) |
| 4 | 3 | Ya (base case) | return 0 | 0 |

Pemanggilan turun dari 1 ke 4, lalu nilai kembali dihitung naik dari 4 ke 1. Hasil akhir: `4 + 6 + 2 = 12`.

```python
print("recursive_sum_linear([4, 6, 2]) =", recursive_sum_linear([4, 6, 2]))
```

### 6. Base case dan recursive case

- **Base case:** `index == len(data)`, artinya semua elemen sudah dilewati dan tidak ada lagi yang dijumlahkan, sehingga fungsi mengembalikan `0`.
- **Recursive case:** `index < len(data)`, yaitu `data[index] + recursive_sum_linear(data, index + 1)`. Masalah diperkecil dengan memajukan `index` satu langkah, sehingga pemanggilan selalu bergerak mendekati base case.

### 7. Time complexity dan space complexity `recursive_sum_linear`

- **Time complexity: O(n).** Ada `n + 1` pemanggilan (n elemen ditambah satu base case), dan tiap pemanggilan hanya melakukan pekerjaan O(1). Karena tidak memakai slicing, tidak ada penyalinan list pada setiap pemanggilan.
- **Space complexity: O(n).** Setiap pemanggilan menunggu hasil pemanggilan berikutnya, sehingga ada `n + 1` frame di *call stack* sekaligus. Tidak ada struktur data tambahan selain stack tersebut.

## Praktik Mandiri 6.2 — Fibonacci dan Memoization

### 1. Implementasi fibonacci_naive dan fibonacci_memo

```python
def fibonacci_naive(n: int) -> int:
    if n <= 1:
        return n
    return fibonacci_naive(n - 1) + fibonacci_naive(n - 2)


def fibonacci_memo(n: int, memo: dict[int, int] | None = None) -> int:
    if memo is None:
        memo = {}
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fibonacci_memo(n - 1, memo) + fibonacci_memo(n - 2, memo)
    return memo[n]
```

### 2. Menguji keduanya untuk n = 5, 10, 20, dan 30

```python
print(f"{'n':<5}{'naive':<10}{'memo':<10}{'sama?'}")
for n in [5, 10, 20, 30]:
    hasil_naive = fibonacci_naive(n)
    hasil_memo = fibonacci_memo(n)
    print(f"{n:<5}{hasil_naive:<10}{hasil_memo:<10}{hasil_naive == hasil_memo}")
```

Kedua versi menghasilkan nilai yang sama. Versi naif tidak diuji untuk `n` yang terlalu besar (misalnya n = 40 ke atas), karena jumlah pemanggilannya sudah mencapai ratusan juta dan sangat lama.

### 3. Menambahkan penghitung jumlah pemanggilan fungsi

```python
hitung = {"naive": 0, "memo": 0}


def fibonacci_naive_hitung(n: int) -> int:
    hitung["naive"] += 1
    if n <= 1:
        return n
    return fibonacci_naive_hitung(n - 1) + fibonacci_naive_hitung(n - 2)


def fibonacci_memo_hitung(n: int, memo: dict[int, int] | None = None) -> int:
    hitung["memo"] += 1
    if memo is None:
        memo = {}
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fibonacci_memo_hitung(n - 1, memo) + fibonacci_memo_hitung(n - 2, memo)
    return memo[n]


def jumlah_panggilan(n: int) -> tuple[int, int]:
    hitung["naive"] = 0
    hitung["memo"] = 0
    fibonacci_naive_hitung(n)
    fibonacci_memo_hitung(n)
    return hitung["naive"], hitung["memo"]


naive_5, memo_5 = jumlah_panggilan(5)
print("n = 5 -> pemanggilan naive:", naive_5, "| pemanggilan memo:", memo_5)
```

Penghitung dipasang pada salinan fungsi (`fibonacci_naive_hitung` dan `fibonacci_memo_hitung`) agar fungsi asli tetap bersih saat diukur waktunya pada langkah 5.

### 4. Membandingkan jumlah pemanggilan untuk n yang sama

```python
print(f"{'n':<5}{'naive':<14}{'memo':<8}{'naive / memo'}")
for n in [5, 10, 20, 30]:
    p_naive, p_memo = jumlah_panggilan(n)
    print(f"{n:<5}{p_naive:<14}{p_memo:<8}{p_naive / p_memo:.1f}x")
```

Versi naif melakukan pemanggilan yang berlipat ganda setiap `n` naik, sedangkan versi memo hanya bertambah sekitar 2 pemanggilan setiap `n` naik 1 (rumusnya `2n - 1`). Pada n = 30, selisihnya sangat jauh.

### 5. Mengukur waktu dengan timeit

```python
import timeit


def rata_rata_ms(fungsi, ulangan: int) -> float:
    total = timeit.timeit(fungsi, number=ulangan)
    return total / ulangan * 1000


print(f"{'n':<5}{'naive (ms)':<16}{'memo (ms)':<16}{'ulangan naive':<16}{'ulangan memo'}")
for n, ulang_naive, ulang_memo in [(20, 20, 20000), (30, 3, 20000)]:
    t_naive = rata_rata_ms(lambda: fibonacci_naive(n), ulang_naive)
    t_memo = rata_rata_ms(lambda: fibonacci_memo(n), ulang_memo)
    print(f"{n:<5}{t_naive:<16.4f}{t_memo:<16.4f}{ulang_naive:<16}{ulang_memo}")
```

Jumlah pengulangan dibuat berbeda: versi naif untuk n = 30 hanya diulang 3 kali karena satu kali jalan saja sudah lama, sedangkan versi memo sangat cepat sehingga diulang 20.000 kali supaya waktunya terukur stabil. Angka waktu bisa berbeda di tiap komputer atau sesi Colab, tetapi perbandingan besarnya tetap sama.

### 6. Mengapa memoization mengubah jumlah pekerjaan secara drastis

Misalnya `fibonacci(30)` memanggil `fibonacci(29)` dan `fibonacci(28)`, lalu `fibonacci(29)` juga memanggil `fibonacci(28)` lagi, dan begitu seterusnya sehingga pohon pemanggilannya bercabang dua dan jumlah pemanggilannya berlipat ganda.

# Praktik Mandiri 6.3
## Praktik Mandiri 6.3 — Analisis Potongan Kode

```python
def proses_a(data):
    return data[-1]


def proses_b(data):
    total = 0
    for x in data:
        total += x
    return total


def proses_c(data):
    count = 0
    for x in data:
        for y in data:
            if x == y:
                count += 1
    return count
```

### 1. Ukuran input n

Untuk ketiga fungsi, ukuran input `n` adalah **banyaknya elemen pada `data`**, yaitu `n = len(data)`.

### 2. Operasi dominan

| Fungsi | Operasi dominan |
|---|---|
| proses_a | akses elemen terakhir `data[-1]` |
| proses_b | penambahan `total += x` di dalam loop |
| proses_c | perbandingan `x == y` di dalam loop bersarang |

### 3. Kompleksitas waktu

| Fungsi | Kompleksitas waktu |
|---|---|
| proses_a | O(1) |
| proses_b | O(n) |
| proses_c | O(n²) |

### 4. Alasan dengan menghitung frekuensi operasi utama

| Fungsi | Operasi utama | Frekuensi | T(n) |
|---|---|---|---|
| proses_a | akses `data[-1]` | 1 kali, berapa pun nilai n | 1 |
| proses_b | `total += x` | loop berjalan n kali, tiap iterasi 1 penambahan | n |
| proses_c | `x == y` | loop luar n kali, loop dalam n kali, jadi n × n | n² |

- `proses_a` hanya mengakses satu posisi tanpa perulangan, sehingga frekuensinya tetap 1 dan kompleksitasnya **O(1)**.
- `proses_b` menelusuri seluruh elemen tepat satu kali, sehingga operasinya bertambah sebanding dengan n dan kompleksitasnya **O(n)**.
- `proses_c` untuk setiap `x` menelusuri lagi seluruh `y`, sehingga perbandingan dilakukan n × n = n² kali. Kompleksitasnya **O(n²)**.

### 5. Ruang tambahan

| Fungsi | Ruang tambahan | Keterangan |
|---|---|---|
| proses_a | O(1) | tidak ada variabel tambahan selain nilai yang dikembalikan |
| proses_b | O(1) | hanya variabel `total` dan `x` |
| proses_c | O(1) | hanya variabel `count`, `x`, dan `y` |

Ruang untuk list `data` itu sendiri tidak dihitung sebagai ruang tambahan karena merupakan input.

### 6. Eksperimen kecil untuk n = 100, 1.000, dan 5.000

`proses_c` hanya dijalankan untuk n = 100 dan n = 1.000. Pada n = 5.000 perbandingannya mencapai 25 juta kali sehingga dilewati, dan waktunya hanya diestimasi dari pola n².

```python
import time


def ukur_ms(fungsi, data: list[int], ulangan: int) -> float:
    mulai = time.perf_counter()
    for _ in range(ulangan):
        fungsi(data)
    selesai = time.perf_counter()
    return (selesai - mulai) / ulangan * 1000


print("Jumlah operasi dominan")
print(f"{'n':<8}{'proses_a':<12}{'proses_b':<12}{'proses_c'}")
for n in [100, 1000, 5000]:
    print(f"{n:<8}{1:<12}{n:<12}{n * n:,}")

print("\nRata-rata waktu per pemanggilan (ms)")
print(f"{'n':<8}{'proses_a':<14}{'proses_b':<14}{'proses_c'}")

waktu_c_1000 = 0.0
for n, ulang_b, ulang_c in [(100, 20000, 200), (1000, 2000, 3), (5000, 400, 0)]:
    data = list(range(n))
    t_a = ukur_ms(proses_a, data, 100000)
    t_b = ukur_ms(proses_b, data, ulang_b)
    if ulang_c > 0:
        t_c = ukur_ms(proses_c, data, ulang_c)
        teks_c = f"{t_c:.4f}"
        if n == 1000:
            waktu_c_1000 = t_c
    else:
        teks_c = f"dilewati (estimasi ~{waktu_c_1000 * 25:.0f} ms)"
    print(f"{n:<8}{t_a:<14.6f}{t_b:<14.4f}{teks_c}")
```

**Kesimpulan eksperimen:** waktu `proses_a` hampir tidak berubah walaupun n membesar (O(1)). Waktu `proses_b` naik kira-kira sebanding dengan n (n naik 5 kali, waktu naik sekitar 5 kali). Waktu `proses_c` naik jauh lebih cepat, kira-kira sebanding dengan n² (n naik 10 kali dari 100 ke 1.000, waktu naik sekitar 100 kali). Pola ini sesuai dengan analisis O(1), O(n), dan O(n²). Angka waktu pastinya bergantung pada komputer atau sesi Colab yang dipakai.

### 1. List terurut berisi 20 bilangan

```python
data = [3, 8, 12, 15, 21, 27, 33, 38, 42, 47,
        51, 56, 60, 65, 71, 78, 84, 89, 93, 99]

print("Jumlah elemen:", len(data))
print("Terurut?     :", data == sorted(data))
```

### 2. Implementasi binary_search_recursive

```python
def binary_search_recursive(data: list[int], target: int, low: int = 0, high: int | None = None) -> int:
    if high is None:
        high = len(data) - 1

    if low > high:
        return -1

    mid = (low + high) // 2

    if data[mid] == target:
        return mid
    elif target < data[mid]:
        return binary_search_recursive(data, target, low, mid - 1)
    else:
        return binary_search_recursive(data, target, mid + 1, high)
```

### 3. Menguji target di awal, tengah, akhir, dan target yang tidak ada

```python
uji = [
    ("awal", 3),
    ("tengah", 47),
    ("akhir", 99),
    ("tidak ada", 50),
]

for keterangan, target in uji:
    print(f"Target {target:<4} ({keterangan:<9}) -> index {binary_search_recursive(data, target)}")
```

Hasil `-1` berarti target tidak ditemukan. Target di awal berada pada index 0, target tengah (47) pada index 9, dan target akhir (99) pada index 19.

### 4. Menambahkan print low, high, dan mid pada setiap pemanggilan

```python
def binary_search_recursive(data: list[int], target: int, low: int = 0, high: int | None = None) -> int:
    if high is None:
        high = len(data) - 1

    if low > high:
        print(f"low={low}, high={high} -> low > high, target tidak ada")
        return -1

    mid = (low + high) // 2
    print(f"low={low}, high={high}, mid={mid}, data[mid]={data[mid]}")

    if data[mid] == target:
        return mid
    elif target < data[mid]:
        return binary_search_recursive(data, target, low, mid - 1)
    else:
        return binary_search_recursive(data, target, mid + 1, high)


for keterangan, target in uji:
    print(f"\n== Target {target} ({keterangan}) ==")
    hasil = binary_search_recursive(data, target)
    print("Hasil index:", hasil)
```

Setiap pemanggilan menyempitkan ruang pencarian: jika target lebih kecil dari `data[mid]`, `high` digeser menjadi `mid - 1`, dan jika lebih besar, `low` digeser menjadi `mid + 1`. Ruang pencarian berkurang kira-kira setengahnya pada tiap pemanggilan.

### 6. Membandingkan jumlah langkah dengan linear search pada data yang sama

```python
def linear_search(data: list[int], target: int) -> tuple[int, int]:
    pemeriksaan = 0
    for i, nilai in enumerate(data):
        pemeriksaan += 1
        if nilai == target:
            return i, pemeriksaan
    return -1, pemeriksaan


def binary_search_hitung(data: list[int], target: int, low: int = 0, high: int | None = None) -> tuple[int, int]:
    if high is None:
        high = len(data) - 1

    if low > high:
        return -1, 0

    mid = (low + high) // 2

    if data[mid] == target:
        return mid, 1
    if target < data[mid]:
        indeks, langkah = binary_search_hitung(data, target, low, mid - 1)
    else:
        indeks, langkah = binary_search_hitung(data, target, mid + 1, high)
    return indeks, langkah + 1


print(f"{'Target':<8}{'Linear (index, pemeriksaan)':<30}{'Binary (index, pemeriksaan)'}")
for target in [3, 33, 47, 99, 50]:
    hasil_linear = linear_search(data, target)
    hasil_binary = binary_search_hitung(data, target)
    print(f"{target:<8}{str(hasil_linear):<30}{hasil_binary}")
```

Satu langkah dihitung sebagai satu elemen yang diperiksa (dibandingkan dengan target). Hasilnya:

- Untuk target 33, 47, 99, dan 50, binary search memakai jauh lebih sedikit langkah daripada linear search. Pada target yang tidak ada (50) atau berada di akhir (99), linear search harus memeriksa seluruh 20 elemen, sedangkan binary search hanya 4 langkah (target 50) dan 5 langkah (target 99).
- Pengecualian ada pada target di awal (3): linear search cukup 1 langkah, sedangkan binary search 4 langkah. Jadi binary search tidak selalu lebih cepat untuk setiap target, tetapi pada kasus terburuk jumlah langkahnya hanya sekitar log₂(n), yaitu paling banyak 5 untuk n = 20, sedangkan linear search bisa sampai n = 20.

### 7. Syarat data sebelum binary search digunakan

1. **Data harus sudah terurut** (naik atau turun, dan arah perbandingan pada kode harus sama dengan arah urutannya). Binary search membuang setengah ruang pencarian berdasarkan perbandingan dengan `data[mid]`, dan keputusan itu hanya benar jika data terurut.
2. **Data harus bisa diakses langsung lewat index** (seperti list atau array), supaya posisi tengah `mid` bisa dihitung dan diakses dalam O(1). Pada linked list hal ini tidak bisa dilakukan.
3. **Elemen harus bisa dibandingkan** satu sama lain (misalnya sama-sama angka), karena algoritma bergantung pada operasi `==` dan `<`.
4. Jika data diubah (ditambah atau dihapus), urutannya harus tetap terjaga sebelum pencarian berikutnya.
