##Praktik Mandiri 4.1 — Operasi Dasar Stack

### 1. Membuat class Stack menggunakan Python list

````python
from __future__ import annotations

from typing import Any


class Stack:
    def __init__(self) -> None:
        self.items: list[Any] = []

    def is_empty(self) -> bool:
        return len(self.items) == 0

    def push(self, data: Any) -> None:
        self.items.append(data)

    def pop(self) -> Any:
        if self.is_empty():
            raise IndexError("Stack kosong, tidak ada elemen yang bisa dikeluarkan")
        return self.items.pop()

    def peek(self) -> Any:
        if self.is_empty():
            raise IndexError("Stack kosong, tidak ada elemen top")
        return self.items[-1]

    def size(self) -> int:
        return len(self.items)
````

### 2. Menambahkan data 10, 20, 30, 40 menggunakan push

````python
s = Stack()
s.push(10)
s.push(20)
s.push(30)
s.push(40)
print("Isi stack (bawah -> atas):", s.items)
````

### 3. Menampilkan top dan size

````python
print("Top :", s.peek())
print("Size:", s.size())
````

# Praktik Mandiri 4.2
##Praktik Mandiri 4.2 — Balanced Brackets

Fungsi di bawah ini memakai class `Stack` dari Praktik Mandiri 4.1.

### 1. Implementasi fungsi is_balanced()

````python
def is_balanced(ekspresi: str) -> bool:
    stack = Stack()
    pasangan = {")": "(", "]": "[", "}": "{"}

    for karakter in ekspresi:
        if karakter in "([{":
            stack.push(karakter)
        else:
            if stack.is_empty() or stack.pop() != pasangan.get(karakter):
                return False

    return stack.is_empty()
````

### 2. Menguji ekspresi: (a+b), ([{}]), ([)], ((a+b), dan {a+[b*c]}

````python
daftar_uji = ["(a+b)", "([{}])", "([)]", "((a+b)", "{a+[b*c]}"]

for e in daftar_uji:
    print(f"{e:<12} -> {is_balanced(e)}")
````

**Catatan hasil:** `(a+b)` dan `{a+[b*c]}` seharusnya seimbang, tetapi hasilnya `False`. Penyebabnya, versi awal ini menganggap setiap karakter yang bukan pembuka sebagai penutup, sehingga huruf dan operator (`a`, `+`, `b`, `*`, `c`) ikut dibandingkan dan membuat pengecekan gagal. Hal ini diperbaiki pada langkah 4.

### 4. Modifikasi fungsi agar mengabaikan karakter selain enam simbol bracket

````python
def is_balanced(ekspresi: str) -> bool:
    stack = Stack()
    pasangan = {")": "(", "]": "[", "}": "{"}

    for karakter in ekspresi:
        if karakter in "([{":
            stack.push(karakter)
        elif karakter in ")]}":
            if stack.is_empty() or stack.pop() != pasangan[karakter]:
                return False
        # karakter selain enam simbol bracket diabaikan

    return stack.is_empty()


for e in daftar_uji:
    print(f"{e:<12} -> {is_balanced(e)}")
````

Sekarang semua hasilnya benar: `(a+b)` dan `{a+[b*c]}` seimbang (`True`), sedangkan `([)]` dan `((a+b)` tidak seimbang (`False`).

### 5. Mengapa hanya membandingkan jumlah pembuka dan penutup tidak cukup

````python
for e in ["([)]", ")(", "(]"]:
    buka = sum(e.count(c) for c in "([{")
    tutup = sum(e.count(c) for c in ")]}")
    print(f"{e:<6} pembuka={buka} penutup={tutup} sama={buka == tutup}  is_balanced={is_balanced(e)}")
````

Menghitung jumlah saja tidak cukup karena **urutan** dan **jenis** pasangannya juga harus benar. Pada `([)]` jumlah pembuka dan penutup sama, tetapi `)` menutup di saat top-nya `[`, sehingga tumpang tindih. Pada `)(` jumlahnya juga sama, tetapi penutup muncul sebelum pembukanya. Stack dibutuhkan karena kurung yang paling terakhir dibuka harus menjadi yang pertama ditutup (LIFO).

### 6. Kompleksitas waktu dan ruang

- **Waktu: O(n)**, karena setiap karakter diperiksa tepat satu kali, dan operasi `push` serta `pop` masing-masing O(1).
- **Ruang: O(n)**, karena pada kasus terburuk (misalnya `((((((`) semua karakter pembuka tersimpan di Stack sekaligus.

# Praktik Mandiri 4.3
## Praktik Mandiri 4.3 — Stack Berbasis Linked List

### 1. Membuat class Node dan LinkedStack

````python
class Node:
    def __init__(self, data: Any) -> None:
        self.data: Any = data
        self.next: Node | None = None


class LinkedStack:
    def __init__(self) -> None:
        self.top: Node | None = None
        self.count: int = 0

    def is_empty(self) -> bool:
        return self.top is None

    def push(self, data: Any) -> None:
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node
        self.count += 1

    def pop(self) -> Any:
        if self.top is None:
            raise IndexError("Stack kosong, tidak ada elemen yang bisa dikeluarkan")
        data = self.top.data
        self.top = self.top.next
        self.count -= 1
        return data

    def peek(self) -> Any:
        if self.top is None:
            raise IndexError("Stack kosong, tidak ada elemen top")
        return self.top.data

    def size(self) -> int:
        return self.count

    def display(self) -> None:
        parts = ["top"]
        current = self.top
        while current is not None:
            parts.append(str(current.data))
            current = current.next
        parts.append("None")
        print(" -> ".join(parts))
````

### 2. Menjalankan push untuk data A, B, C

````python
ls = LinkedStack()

ls.push("A")
ls.display()

ls.push("B")
ls.display()

ls.push("C")
ls.display()
````

### 3. Gambar hubungan top dan next setelah setiap push

```
Setelah push("A"):
top -> [A | next=None]

Setelah push("B"):
top -> [B | next] -> [A | next=None]

Setelah push("C"):
top -> [C | next] -> [B | next] -> [A | next=None]
```

Setiap push membuat node baru, `next`-nya diarahkan ke node top yang lama, lalu `top` dipindah ke node baru.

### 4. Menjalankan pop sekali dan menggambar ulang strukturnya

````python
print("Nilai yang keluar:", ls.pop())
ls.display()
````

```
Setelah pop():
top -> [B | next] -> [A | next=None]

Node C sudah lepas dari rantai karena top pindah ke node B.
```

### 5. Menguji peek dan size

````python
print("Peek:", ls.peek())
print("Size:", ls.size())
````

### 6. Perbandingan push/pop LinkedStack dengan insert/delete awal Singly Linked List Pertemuan 3

| | Singly Linked List (Pertemuan 3) | LinkedStack |
|---|---|---|
| Menambah di awal | `insert_beginning`:<br>`new_node.next = self.head`<br>`self.head = new_node` | `push`:<br>`new_node.next = self.top`<br>`self.top = new_node` |
| Menghapus di awal | `delete` (kasus head):<br>`self.head = self.head.next` | `pop`:<br>`data = self.top.data`<br>`self.top = self.top.next` |

Logikanya sama persis, hanya nama penunjuknya yang berbeda (`head` pada Singly Linked List, `top` pada LinkedStack). Perbedaannya, `pop` selalu menghapus node paling atas tanpa mencari nilai dan mengembalikan datanya, sedangkan `delete` pada Singly Linked List menghapus berdasarkan nilai sehingga perlu menelusuri list dan hanya mengembalikan `True` atau `False`.

### 7. Mengapa keduanya dapat O(1)

`push` dan `pop` pada LinkedStack serta `insert_beginning` dan penghapusan head pada Singly Linked List selalu bekerja di ujung depan yang langsung ditunjuk oleh `top` (atau `head`). Tidak ada perulangan untuk menelusuri node, hanya membuat satu node dan mengubah beberapa penunjuk. Jumlah langkahnya tetap sama berapa pun banyaknya node, sehingga kompleksitasnya **O(1)**.
