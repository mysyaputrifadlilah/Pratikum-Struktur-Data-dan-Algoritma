```python
## 1. Membuat List

nilai = [12, 45, 7, 33, 90, 21, 58, 3, 76, 40]
print(nilai)
print(type(nilai))
```

```python
## 2. Mengakses Elemen

print("Elemen pertama (indeks 0) :", nilai[0])
print("Elemen ke-5 (indeks 4)    :", nilai[4])
print("Elemen terakhir (indeks -1):", nilai[-1])
```

```python
## 3. Mengubah Nilai pada Indeks Tertentu

print("Sebelum :", nilai)
nilai[3] = 99   # ubah 33 menjadi 99
print("Sesudah :", nilai)
```

```python
## 4. Menambah Nilai dengan `append()`

nilai.append(100)
nilai.append(55)
print(nilai)
```

```python
## 5. Menyisipkan Nilai dengan `insert()`

nilai.insert(2, 25)
print(nilai)
```

```python
## 6. Menghapus Elemen dengan `pop()` dan `remove()`

nilai = [12, 45, 7, 33, 90, 21, 58, 3, 76, 40]
dihapus = nilai.pop(3)
print('Dihapus:', dihapus)
print('nilai:', nilai)

nilai = [12, 45, 7, 33, 90, 21, 58, 3, 76, 40]
nilai.remove(90)
print(nilai)
```

```python
## 7. Traversal dengan `enumerate()`

for indeks, isi in enumerate(nilai):
    print(f"Indeks {indeks} : {isi}")
```

## 8. Catatan: Operasi yang Menyebabkan Pergeseran Elemen

- insert(i, x)`menggeser semua elemen dari indeks `i` ke kanan (indeks bertambah 1), karena harus menyediakan tempat untuk nilai baru.
- remove(x)` dan `pop(i)`(jika bukan elemen terakhir) menggeser elemen setelah posisi yang dihapus ke kiri (indeks berkurang 1), karena list tidak boleh punya "lubang".
- pop()` tanpa argumen** dan`append()` tidak menggeser elemen lain karena hanya bekerja di ujung list.
- Mengubah nilai lewat indeks (`nilai[3] = 99`) tidak menggeser apa pun.
- Karena pergeseran ini, `insert` dan `remove` di bagian awal/tengah list lebih lambat (O(n)) dibanding `append` dan `pop()` di ujung (O(1)).

7. mengapa worst case Linear Search adalah O(n).

Pada Worst Case, target berada di elemen terakhir atau tidak ditemukan, sehingga semua elemen harus diperiksa satu per satu. Jika jumlah data adalah n, maka jumlah pemeriksaan bisa mencapai n kali. Karena jumlah pemeriksaan bertambah sebanding dengan jumlah data, kompleksitas waktunya adalah O(n). Jadi, semakin banyak data yang dicari, semakin banyak pula pemeriksaan yang diperlukan.

Praktik Mandiri 2.3 — Binary Search

```python
# 1. Membuat list minimal 20 angka
data = [45, 12, 78, 3, 56, 89, 23, 67, 34, 91,
        10, 50, 72, 5, 29, 83, 41, 16, 60, 95]

# Mengurutkan data
data = sorted(data)

print("Data terurut:")
print(data)
```

```python
def binary_search_trace(data, target):
    left = 0
    right = len(data) - 1
    jumlah = 0

    while left <= right:
        mid = (left + right) // 2
        jumlah = jumlah + 1

        print("left =", left, "right =", right, "mid =", mid)

        if data[mid] == target:
            return mid, jumlah

        elif data[mid] < target:
            left = mid + 1

        else:
            right = mid - 1

    return -1, jumlah

    # 2. Target di sekitar tengah
print()
print("TARGET TENGAH")

target = 50
index, jumlah = binary_search_trace(data, target)

print("Target =", target)
print("Indeks =", index)
print("Jumlah pemeriksaan =", jumlah)
```

```python
# 3. Target dekat awal
print()
print("TARGET DEKAT AWAL")

target = 5
index, jumlah = binary_search_trace(data, target)

print("Target =", target)
print("Indeks =", index)
print("Jumlah pemeriksaan =", jumlah)

# Target dekat akhir
print()
print("TARGET DEKAT AKHIR")

target = 91
index, jumlah = binary_search_trace(data, target)

print("Target =", target)
print("Indeks =", index)
print("Jumlah pemeriksaan =", jumlah)

```

```python
# 4. Target yang tidak ada
print()
print("TARGET TIDAK ADA")

target = 70
index, jumlah = binary_search_trace(data, target)

print("Target =", target)
print("Indeks =", index)
print("Jumlah pemeriksaan =", jumlah)
```

```python
# 7. Membuat fungsi Linear Search
def linear_search(data, target):
    jumlah = 0

    for i in range(len(data)):
        jumlah = jumlah + 1

        if data[i] == target:
            return i, jumlah

    return -1, jumlah

    # Perbandingan Binary Search dan Linear Search
print()
print("PERBANDINGAN BINARY SEARCH DAN LINEAR SEARCH")

target = 91

index_binary, jumlah_binary = binary_search_trace(data, target)
index_linear, jumlah_linear = linear_search(data, target)

print("Target =", target)
print("Binary Search:")
print("Indeks =", index_binary)
print("Jumlah pemeriksaan =", jumlah_binary)

print("Linear Search:")
print("Indeks =", index_linear)
print("Jumlah pemeriksaan =", jumlah_linear)
```
