##Praktik Mandiri 5.1 — Queue Dasar

### 1. Membuat class Queue berbasis collections.deque

````python
from __future__ import annotations

from collections import deque
from typing import Any


class Queue:
    def __init__(self) -> None:
        self.items: deque[Any] = deque()

    def is_empty(self) -> bool:
        return len(self.items) == 0

    def enqueue(self, data: Any) -> None:
        self.items.append(data)

    def dequeue(self) -> Any:
        if self.is_empty():
            raise IndexError("Queue kosong, tidak ada elemen yang bisa dikeluarkan")
        return self.items.popleft()

    def peek(self) -> Any:
        if self.is_empty():
            raise IndexError("Queue kosong, tidak ada elemen front")
        return self.items[0]

    def size(self) -> int:
        return len(self.items)
````

### 2. Memasukkan data 10, 20, 30, 40, 50

````python
q = Queue()
for nilai in [10, 20, 30, 40, 50]:
    q.enqueue(nilai)

print("Isi queue (front -> rear):", list(q.items))
````

### 3. Menampilkan front dan size

````python
print("Front:", q.peek())
print("Size :", q.size())
````

### 4. Melakukan dequeue dua kali dan mencatat nilai yang keluar

````python
print("dequeue pertama:", q.dequeue())
print("dequeue kedua  :", q.dequeue())
print("Isi queue      :", list(q.items))
````

**Nilai yang keluar:** 10, lalu 20. Nilai yang masuk paling awal keluar paling awal.

### 5. Menambahkan 60 dan 70

### 6. Mengeluarkan seluruh elemen dengan loop sampai Queue kosong

````python
while not q.is_empty():
    print("dequeue:", q.dequeue())

print("Queue kosong?", q.is_empty())
````

### 7. Menguji dequeue dan peek ketika Queue kosong

````python
try:
    q.dequeue()
except IndexError as e:
    print("dequeue ->", "IndexError:", e)

try:
    q.peek()
except IndexError as e:
    print("peek    ->", "IndexError:", e)
````

Saat Queue kosong, tidak ada elemen untuk dikeluarkan maupun dilihat. Keduanya mengecek `is_empty()` terlebih dahulu, lalu melempar `IndexError` dengan pesan yang jelas. Error ditangkap dengan `try ... except`, sehingga program tidak berhenti.

### 9. Bukti bahwa urutan keluar mengikuti FIFO

````python
uji = Queue()
masuk: list[int] = []
keluar: list[int] = []

for nilai in [10, 20, 30, 40, 50]:
    uji.enqueue(nilai)
    masuk.append(nilai)

for _ in range(2):
    keluar.append(uji.dequeue())

for nilai in [60, 70]:
    uji.enqueue(nilai)
    masuk.append(nilai)

while not uji.is_empty():
    keluar.append(uji.dequeue())

print("Urutan masuk :", masuk)
print("Urutan keluar:", keluar)
print("Sama persis? :", masuk == keluar)
````

Urutan keluar (10, 20, 30, 40, 50, 60, 70) sama persis dengan urutan masuk, walaupun proses `enqueue` dan `dequeue` bergantian. Data yang masuk lebih dulu selalu keluar lebih dulu, itulah prinsip **FIFO** (First In, First Out).

# Praktik Mandiri 5.2
##Praktik Mandiri 5.2 — Circular Queue

### 1. Membuat CircularQueue dengan kapasitas 4

````python
class CircularQueue:
    def __init__(self, capacity: int) -> None:
        self.capacity: int = capacity
        self.items: list[Any] = [None] * capacity
        self.front: int = 0
        self.rear: int = -1
        self.count: int = 0

    def is_empty(self) -> bool:
        return self.count == 0

    def is_full(self) -> bool:
        return self.count == self.capacity

    def enqueue(self, data: Any) -> bool:
        if self.is_full():
            return False
        self.rear = (self.rear + 1) % self.capacity
        self.items[self.rear] = data
        self.count += 1
        return True

    def dequeue(self) -> Any:
        if self.is_empty():
            raise IndexError("Queue kosong, tidak ada elemen yang bisa dikeluarkan")
        data = self.items[self.front]
        self.items[self.front] = None
        self.front = (self.front + 1) % self.capacity
        self.count -= 1
        return data


def tampilkan(cq: CircularQueue, keterangan: str = "") -> None:
    print(f"{keterangan:<20} items={cq.items}, front={cq.front}, rear={cq.rear}, count={cq.count}")


cq = CircularQueue(4)
tampilkan(cq, "Awal")
````

### 2. Enqueue A, B, C, D dan cek kondisi penuh

````python
for huruf in ["A", "B", "C", "D"]:
    cq.enqueue(huruf)
    tampilkan(cq, f"enqueue({huruf})")

print("\nQueue penuh?", cq.is_full())
````

### 7. Keadaan array fisik dan urutan Queue logis

**Setelah dua kali dequeue** (`front = 2`, `rear = 3`, `count = 2`):

```
Indeks      0     1     2     3
Isi       [   ] [   ] [ C ] [ D ]
                        ^     ^
                      front  rear
```
Urutan logis (front -> rear): `C -> D`

**Setelah enqueue E dan F** (`front = 2`, `rear = 1`, `count = 4`):

```
Indeks      0     1     2     3
Isi       [ E ] [ F ] [ C ] [ D ]
                  ^     ^
                 rear  front
```
Urutan logis (front -> rear): `C -> D -> E -> F`

Pada array fisik, E dan F berada di indeks 0 dan 1 (sebelum C dan D), tetapi secara logis keduanya antre paling belakang.

### 8. Fungsi operasi modulo pada implementasi

Operasi modulo (`%`) membuat indeks berputar kembali ke awal array setelah mencapai ujung:

- `rear = (rear + 1) % capacity` menentukan posisi enqueue berikutnya.
- `front = (front + 1) % capacity` menentukan posisi dequeue berikutnya.

Contoh: saat `rear = 3` dan kapasitas 4, maka `(3 + 1) % 4 = 0`. Rear kembali ke indeks 0, sehingga slot kosong di awal array yang sudah ditinggalkan oleh dequeue bisa dipakai lagi tanpa menggeser elemen. Tanpa modulo, indeks akan melewati batas array.

### 9. Kompleksitas enqueue dan dequeue

- **enqueue: O(1)** waktu, karena hanya menghitung indeks `rear` baru dan mengisi satu slot.
- **dequeue: O(1)** waktu, karena hanya mengambil elemen di `front` dan menggeser indeks `front`, tanpa menggeser elemen lain.
- **Ruang:** O(n) untuk array berkapasitas tetap `n` (di sini n = 4), dan setiap operasi hanya memakai ruang tambahan O(1).

# Praktik Mandiri 5.3
## Praktik Mandiri 5.3 — Deque dan Palindrome

### 1. Implementasi fungsi is_palindrome()

````python
def is_palindrome(teks: str) -> bool:
    d = deque(teks)

    while len(d) > 1:
        if d.popleft() != d.pop():
            return False

    return True
````

### 2. Menguji kata: level, katak, radar, python, algoritma

````python
for kata in ["level", "katak", "radar", "python", "algoritma"]:
    print(f"{kata:<10} -> {is_palindrome(kata)}")
````

Hasilnya: `level`, `katak`, dan `radar` adalah palindrome (`True`), sedangkan `python` dan `algoritma` bukan (`False`).

### 3. Menguji kalimat yang mengandung spasi atau tanda baca setelah normalisasi

````python
kalimat_uji = [
    "Kasur ini rusak",
    "A man, a plan, a canal: Panama",
    "Belajar struktur data",
]

print("Sebelum normalisasi:")
for kalimat in kalimat_uji:
    print(f"{kalimat:<32} -> {is_palindrome(kalimat)}")
````

Tanpa normalisasi, `Kasur ini rusak` dan `A man, a plan, a canal: Panama` terbaca bukan palindrome, padahal seharusnya palindrome. Penyebabnya, spasi, tanda baca, dan perbedaan huruf besar-kecil ikut dibandingkan. Karena itu teks dinormalisasi dulu: hanya huruf dan angka yang dipakai, lalu diubah menjadi huruf kecil.

````python
def normalisasi(teks: str) -> str:
    return "".join(karakter.lower() for karakter in teks if karakter.isalnum())


def is_palindrome(teks: str) -> bool:
    d = deque(normalisasi(teks))

    while len(d) > 1:
        if d.popleft() != d.pop():
            return False

    return True


print("Setelah normalisasi:")
for kalimat in kalimat_uji:
    print(f"{kalimat:<32} -> {is_palindrome(kalimat)}")
````

Sekarang hasilnya benar: `Kasur ini rusak` dan `A man, a plan, a canal: Panama` adalah palindrome (`True`), sedangkan `Belajar struktur data` bukan (`False`).

### 4. Dry run pasangan karakter untuk satu contoh palindrome: `radar`

| Iterasi | Isi Deque sebelum dibandingkan | Karakter kiri (popleft) | Karakter kanan (pop) | Sama? | Sisa Deque |
|---|---|---|---|---|---|
| 1 | [r, a, d, a, r] | r | r | Ya | [a, d, a] |
| 2 | [a, d, a] | a | a | Ya | [d] |
| - | [d], panjang 1, loop berhenti | - | - | - | [d] |

Semua pasangan sama dan tersisa satu karakter di tengah, maka hasilnya `True` (palindrome).

### 5. Mengubah fungsi agar menampilkan pasangan kiri-kanan pada setiap iterasi

````python
def is_palindrome(teks: str, tampilkan: bool = False) -> bool:
    d = deque(normalisasi(teks))
    iterasi = 1

    while len(d) > 1:
        kiri = d.popleft()
        kanan = d.pop()
        if tampilkan:
            print(f"Iterasi {iterasi}: kiri = {kiri!r}, kanan = {kanan!r}, sisa = {list(d)}")
        if kiri != kanan:
            return False
        iterasi += 1

    return True


print("radar:")
print("Hasil:", is_palindrome("radar", tampilkan=True))

print("\nalgoritma:")
print("Hasil:", is_palindrome("algoritma", tampilkan=True))
````

Pada `radar` semua pasangan sama, sedangkan pada `algoritma` iterasi 2 sudah menemukan pasangan `'l'` dan `'m'` yang berbeda, sehingga fungsi langsung berhenti dengan hasil `False`.

### 6. Mengapa Deque lebih natural daripada Queue satu arah

Algoritma palindrome membandingkan karakter pertama dengan karakter terakhir, lalu karakter kedua dengan kedua dari belakang, dan seterusnya bergerak ke tengah. Artinya data harus diambil dari **kedua ujung**. Deque memang mendukung hapus di depan (`popleft`) dan di belakang (`pop`), masing-masing O(1). Queue satu arah hanya bisa mengeluarkan data dari depan, sehingga untuk mengambil karakter terakhir kita harus mengeluarkan semua karakter sebelumnya terlebih dahulu atau memakai struktur tambahan. Itu membuat kodenya lebih rumit dan kurang efisien.

### 7. Kompleksitas waktu dan ruang

Misalkan panjang teks adalah `n`.

- **Waktu: O(n)**. Normalisasi memeriksa `n` karakter, pembuatan deque memerlukan O(n), dan loop berjalan paling banyak `n/2` iterasi dengan setiap `popleft` dan `pop` bernilai O(1).
- **Ruang: O(n)**, karena string hasil normalisasi dan deque masing-masing menyimpan sampai `n` karakter.
