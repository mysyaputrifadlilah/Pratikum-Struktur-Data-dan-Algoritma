## Tugas 1 — Linear Search

Fungsi `linear_search_nim(data, target)` memeriksa kode mahasiswa dari elemen pertama sampai akhir.
Fungsi mengembalikan **(indeks, jumlah_perbandingan)**. Jika tidak ditemukan, indeks = `-1`.

```python
def linear_search_nim(data, target):
    jumlah_banding = 0                      # penghitung perbandingan
    for i in range(len(data)):
        jumlah_banding += 1                 # 1 kali bandingkan kode data[i] dengan target
        if data[i][0] == target:
            return i, jumlah_banding        # ketemu -> kembalikan indeks
    return -1, jumlah_banding               # sampai akhir tidak ketemu
```

## Tugas 3 — Merge dan Merge Sort Rekursif
- `merge(left, right)` menggabungkan dua list terurut menjadi satu list terurut berdasarkan kode (ascending).
- `merge_sort_nim(data)` membagi list dua secara **rekursif** sampai 1 elemen (basis), lalu menggabungkannya dengan `merge`. Fungsi membuat list baru, sehingga data awal tidak berubah.

```python
def merge(left, right):
    hasil = []
    i = 0                                   # pointer list kiri
    j = 0                                   # pointer list kanan

    # bandingkan elemen terdepan kedua list, ambil yang kodenya lebih kecil
    while i < len(left) and j < len(right):
        if left[i][0] <= right[j][0]:       # <= agar stabil (urutan sama tetap)
            hasil.append(left[i])
            i += 1
        else:
            hasil.append(right[j])
            j += 1

    # salah satu list habis -> sisa list lain langsung ditambahkan
    while i < len(left):
        hasil.append(left[i])
        i += 1
    while j < len(right):
        hasil.append(right[j])
        j += 1

    return hasil


def merge_sort_nim(data):
    # basis rekursi: 0 atau 1 elemen sudah terurut
    if len(data) <= 1:
        return data[:]                      # kembalikan salinan

    mid = len(data) // 2
    left = merge_sort_nim(data[:mid])       # urutkan bagian kiri (rekursif)
    right = merge_sort_nim(data[mid:])      # urutkan bagian kanan (rekursif)
    return merge(left, right)               # gabungkan hasilnya
```

## Tugas 4 — Data Sebelum dan Sesudah Pengurutan
Data awal ternyata sudah urut, jadi ditambahkan **uji tambahan** memakai data yang dibalik (`data[::-1]`) untuk membuktikan Merge Sort benar-benar bekerja.

```python
def tampil(judul, arr):
    print(judul)
    print(f"{'No':<4}{'Kode':<8}{'Nama':<10}{'Nilai':<6}")
    for no, (kode, nama, nilai) in enumerate(arr):
        print(f"{no:<4}{kode:<8}{nama:<10}{nilai:<6}")
    print()

def sudah_urut(arr):                        # cek urutan secara manual (perulangan)
    for k in range(len(arr) - 1):
        if arr[k][0] > arr[k + 1][0]:
            return False
    return True

data_awal_salinan = [baris[:] for baris in data]   # untuk membuktikan data asli tidak berubah

data_urut = merge_sort_nim(data)

tampil("DATA SEBELUM PENGURUTAN", data)
tampil("DATA SESUDAH PENGURUTAN (data_urut)", data_urut)
print("Data asli tidak berubah :", data == data_awal_salinan)
print("data_urut sudah urut    :", sudah_urut(data_urut))

# --- pengujian tambahan: data dibalik ---
data_balik = data[::-1]
hasil_balik = merge_sort_nim(data_balik)
print()
print("Kode data dibalik :", [x[0] for x in data_balik])
print("Kode hasil sort   :", [x[0] for x in hasil_balik])
print("Hasil sama dengan data_urut :", hasil_balik == data_urut)
```

## Tugas 5 — Binary Search
`binary_search_nim(data_urut, target)` mencari pada data yang **sudah terurut** dan mengembalikan **(indeks, jumlah_perbandingan)**. Dihitung **1 perbandingan per iterasi** (target vs kode tengah).

```python
def binary_search_nim(data_urut, target):
    low = 0
    high = len(data_urut) - 1
    jumlah_banding = 0

    while low <= high:
        mid = (low + high) // 2
        jumlah_banding += 1                 # 1 perbandingan target vs kode tengah
        kode_mid = data_urut[mid][0]

        if kode_mid == target:
            return mid, jumlah_banding      # ketemu
        elif kode_mid < target:
            low = mid + 1                   # target ada di kanan
        else:
            high = mid - 1                  # target ada di kiri

    return -1, jumlah_banding               # tidak ketemu
```

## Tugas 7 — Trace / Dry Run
### (a) Satu proses Merge Sort pada 6 data pertama
`BAGI` menunjukkan batas indeks tiap pembagian, `GABUNG` menunjukkan hasil penggabungan.

```python
def merge_sort_trace(arr, awal=0, depth=0):
    indent = "    " * depth
    kode = [x[0] for x in arr]
    print(f"{indent}BAGI   idx[{awal}..{awal + len(arr) - 1}] : {kode}")

    if len(arr) <= 1:
        print(f"{indent}BASIS  1 elemen, langsung dikembalikan : {kode}")
        return arr[:]

    mid = len(arr) // 2
    print(f"{indent}       mid = {mid} -> kiri {kode[:mid]} | kanan {kode[mid:]}")
    left = merge_sort_trace(arr[:mid], awal, depth + 1)
    right = merge_sort_trace(arr[mid:], awal + mid, depth + 1)
    hasil = merge(left, right)
    print(f"{indent}GABUNG {[x[0] for x in left]} + {[x[0] for x in right]} -> {[x[0] for x in hasil]}")
    return hasil

print("TRACE MERGE SORT — 6 data pertama")
print("=" * 60)
_ = merge_sort_trace(data[:6])
```

### (b) Binary Search untuk `target_nim`

```python
def binary_search_trace(data_urut, target):
    low, high, banding = 0, len(data_urut) - 1, 0
    print(f"TRACE BINARY SEARCH — target = {target}")
    print("=" * 60)
    while low <= high:
        mid = (low + high) // 2
        banding += 1
        kode_mid = data_urut[mid][0]
        print(f"Iterasi {banding}: low={low}, high={high}, mid={mid}, kode[mid]={kode_mid}")
        if kode_mid == target:
            print(f"           {kode_mid} == {target} -> DITEMUKAN di indeks {mid}")
            return mid, banding
        elif kode_mid < target:
            print(f"           {kode_mid} < {target} -> cari di kanan, low = {mid + 1}")
            low = mid + 1
        else:
            print(f"           {kode_mid} > {target} -> cari di kiri, high = {mid - 1}")
            high = mid - 1
    print("Tidak ditemukan")
    return -1, banding

_ = binary_search_trace(data_urut, target_nim)
```

## Tugas 9 — Analisis Kompleksitas

| Algoritma | Best case | Average / Worst case | Ruang tambahan |
|---|---|---|---|
| Linear Search | O(1) | O(n) | O(1) |
| Binary Search | O(1) | O(log n) | O(1) |
| Merge Sort | O(n log n) | O(n log n) | O(n) |

- **Linear Search:** memeriksa satu per satu. Hasil uji: `TI009` → 9 perbandingan, `TI999` (tidak ada) → 12 (= n).
- **Binary Search:** tiap iterasi membuang separuh data, maksimal ⌊log₂ n⌋ + 1 = 4 perbandingan untuk n = 12. Hasil uji: `TI009` → 2, `TI999` → 4.
- **Merge Sort:** rekursi sedalam log n level, tiap level memproses total n elemen pada `merge`, sehingga O(n log n) untuk semua kasus.
- **Ruang tambahan Merge Sort:** `merge` membuat list baru dan pembagian membuat salinan (slicing), sehingga O(n). Stack rekursi hanya O(log n).
- **Kenapa Binary Search butuh data terurut:** separuh data dibuang hanya berdasarkan elemen tengah. Jika `kode[mid] < target`, semua elemen di kirinya pasti lebih kecil juga. Hal ini hanya benar jika data urut.
- **Kapan sorting layak:** sorting O(n log n) dibayar sekali, lalu tiap pencarian turun dari O(n) menjadi O(log n). Layak jika data dicari berkali-kali dan jarang berubah (untuk n = 12, perkiraan kasarnya mulai untung di sekitar 6 pencarian). Jika hanya dicari sekali, Linear Search lebih murah.

**Logika program:** (1) Linear Search menelusuri list dari indeks 0 sambil menghitung perbandingan. (2) Merge Sort membagi list rekursif, lalu `merge` menggabungkan dua list terurut dengan membandingkan elemen terdepan masing-masing. (3) Binary Search mengecek elemen tengah lalu mempersempit `low`–`high` ke kiri atau kanan.

# SOAL 2 — Sistem Antrean Layanan dengan Undo Menggunakan Linked Queue dan Stack

## Tugas 1 — Implementasi Class dan Uji Operasi Dasar
- `Node`: atribut `data` dan `next`.
- `LinkedQueue`: **singly linked list** dengan `front`, `rear`, `count` (tanpa `deque`, `queue.Queue`, atau `list.pop(0)`).
- `Stack`: list Python sebagai penyimpanan internal, semua operasi dibungkus class buatan sendiri.
- `dequeue` pada Queue kosong dan `pop` pada Stack kosong menampilkan pesan dan mengembalikan `None` (tidak error).

```python
class Node:
    def __init__(self, data):
        self.data = data        # isi node (ID layanan)
        self.next = None        # referensi ke node berikutnya


class LinkedQueue:
    """Queue (FIFO) berbasis singly linked list: front = kepala, rear = ekor."""

    def __init__(self):
        self.front = None       # node paling depan (tempat dequeue)
        self.rear = None        # node paling belakang (tempat enqueue)
        self.count = 0          # jumlah elemen

    def is_empty(self):
        return self.count == 0

    def size(self):
        return self.count

    def enqueue(self, data):
        """Tambah di belakang (rear) -> O(1) karena ada referensi rear."""
        node = Node(data)
        if self.is_empty():
            self.front = node
            self.rear = node
        else:
            self.rear.next = node
            self.rear = node
        self.count += 1

    def enqueue_front(self, data):
        """Tambah di depan (front) -> dipakai untuk UNDO. O(1)."""
        node = Node(data)
        if self.is_empty():
            self.front = node
            self.rear = node
        else:
            node.next = self.front
            self.front = node
        self.count += 1

    def dequeue(self):
        """Ambil dari depan (front). Jika kosong: tampilkan pesan, kembalikan None."""
        if self.is_empty():
            print("[PESAN] Queue kosong, dequeue tidak dapat dilakukan.")
            return None
        node = self.front
        self.front = node.next
        if self.front is None:      # queue menjadi kosong -> rear ikut None
            self.rear = None
        node.next = None
        self.count -= 1
        return node.data

    def peek(self):
        """Lihat elemen front tanpa menghapus."""
        if self.is_empty():
            print("[PESAN] Queue kosong, tidak ada elemen untuk dilihat.")
            return None
        return self.front.data

    def display(self):
        if self.is_empty():
            print("Queue (front -> rear)       : [kosong]")
            return
        isi = []
        cur = self.front
        while cur is not None:
            isi.append(str(cur.data))
            cur = cur.next
        print("Queue (front -> rear)       :", " -> ".join(isi))


class Stack:
    """Stack (LIFO) dengan Python list sebagai penyimpanan internal."""

    def __init__(self):
        self._items = []

    def is_empty(self):
        return len(self._items) == 0

    def size(self):
        return len(self._items)

    def push(self, data):
        self._items.append(data)

    def pop(self):
        """Ambil elemen teratas. Jika kosong: tampilkan pesan, kembalikan None."""
        if self.is_empty():
            print("[PESAN] Stack kosong, pop tidak dapat dilakukan.")
            return None
        return self._items.pop()

    def peek(self):
        """Lihat elemen teratas tanpa menghapus."""
        if self.is_empty():
            print("[PESAN] Stack kosong, tidak ada elemen untuk dilihat.")
            return None
        return self._items[-1]

    def display(self):
        if self.is_empty():
            print("Stack riwayat (bawah -> atas): [kosong]")
        else:
            print("Stack riwayat (bawah -> atas):", " | ".join(str(x) for x in self._items), " <- top")
```

### Uji Operasi Dasar (sebelum simulasi utama)
Memakai objek terpisah (`uji_q`, `uji_s`) agar tidak mempengaruhi simulasi.

```python
print("=== UJI LinkedQueue ===")
uji_q = LinkedQueue()
print("is_empty awal :", uji_q.is_empty(), "| size =", uji_q.size())

uji_q.enqueue("X1")
uji_q.enqueue("X2")
uji_q.enqueue("X3")
print("Setelah enqueue X1, X2, X3:")
uji_q.display()

uji_q.enqueue_front("X0")
print("Setelah enqueue_front X0:")
uji_q.display()

print("peek    :", uji_q.peek())
print("dequeue :", uji_q.dequeue())
uji_q.display()
print("front =", uji_q.front.data, "| rear =", uji_q.rear.data, "| count =", uji_q.count)
print("size    :", uji_q.size(), "| is_empty:", uji_q.is_empty())

print()
print("=== UJI Stack ===")
uji_s = Stack()
print("is_empty awal :", uji_s.is_empty(), "| size =", uji_s.size())
uji_s.push("S1")
uji_s.push("S2")
uji_s.push("S3")
print("Setelah push S1, S2, S3:")
uji_s.display()
print("peek    :", uji_s.peek())
print("pop     :", uji_s.pop())
uji_s.display()
print("size    :", uji_s.size(), "| is_empty:", uji_s.is_empty())
```

## Tugas 2 — Simulasi 12 Operasi
ID dibentuk dari `d = 8`: A = `MHS-801`, ..., F = `MHS-806`.
- **DATANG X** → `enqueue(X)`
- **LAYANI** → `dequeue()` dari front; jika berhasil, ID di-`push` ke Stack riwayat
- **UNDO** → `pop()` dari Stack, lalu `enqueue_front(ID)`

```python
NIM = "411252028"
d = int(NIM[-1])
A = f'MHS-{d}01'
B = f'MHS-{d}02'
C = f'MHS-{d}03'
D = f'MHS-{d}04'
E = f'MHS-{d}05'
F = f'MHS-{d}06'


def datang(queue, x):
    queue.enqueue(x)


def layani(queue, stack):
    id_layanan = queue.dequeue()            # ambil dari front
    if id_layanan is not None:              # jika berhasil -> catat ke riwayat
        stack.push(id_layanan)
    return id_layanan


def undo(queue, stack):
    id_layanan = stack.pop()                # ambil riwayat terakhir
    if id_layanan is not None:              # jika ada -> kembalikan ke depan antrean
        queue.enqueue_front(id_layanan)
    return id_layanan


queue = LinkedQueue()
stack = Stack()

urutan = [
    ("DATANG", A), ("DATANG", B), ("DATANG", C), ("LAYANI", None),
    ("DATANG", D), ("LAYANI", None), ("DATANG", E), ("UNDO", None),
    ("LAYANI", None), ("LAYANI", None), ("UNDO", None), ("DATANG", F),
]

print("SIMULASI ANTREAN LAYANAN (d =", d, ")")
print("=" * 70)
for no, (op, x) in enumerate(urutan, start=1):
    if op == "DATANG":
        datang(queue, x)
        ket = f"DATANG {x}"
    elif op == "LAYANI":
        hasil = layani(queue, stack)
        ket = f"LAYANI  -> melayani {hasil}"
    else:
        hasil = undo(queue, stack)
        ket = f"UNDO    -> {hasil} dikembalikan ke depan antrean"

    print(f"Langkah {no:>2} : {ket}")
    queue.display()
    stack.display()
    print("-" * 70)
```

## Tugas 4 — Pengujian Edge Case
(1) `dequeue` pada Queue kosong dan (2) `pop`/UNDO pada Stack riwayat kosong. Program tidak berhenti; pesan ditampilkan dan `None` dikembalikan.

```python
q_kosong = LinkedQueue()
s_kosong = Stack()

print("EDGE CASE 1 : dequeue pada Queue kosong")
hasil1 = q_kosong.dequeue()
print("Nilai kembalian :", hasil1)
print("Size Queue      :", q_kosong.size())
print()

print("EDGE CASE 2 : UNDO ketika Stack riwayat kosong")
hasil2 = undo(q_kosong, s_kosong)
print("Nilai kembalian :", hasil2)
print("pop() langsung pada Stack kosong ->", s_kosong.pop())
print("Size Queue      :", q_kosong.size(), "| Size Stack :", s_kosong.size())
print()

print("EDGE CASE 3 (tambahan) : LAYANI pada antrean kosong lewat alur simulasi")
hasil3 = layani(q_kosong, s_kosong)
print("Nilai kembalian :", hasil3, "| Stack tetap kosong :", s_kosong.is_empty())
print()

print("Program tetap berjalan sampai akhir tanpa error.")
```

## Tugas 5 — FIFO pada Queue dan LIFO pada Stack
- **FIFO:** A, B, C datang berurutan. Langkah 4 melayani **A** (paling awal), langkah 6 melayani **B**. `enqueue` menambah di rear, `dequeue` mengambil dari front.
- **LIFO:** setelah langkah 6, Stack berisi A, B. UNDO langkah 8 meng-`pop` **B** (paling akhir masuk), bukan A. UNDO langkah 11 meng-`pop` **C** (paling akhir dilayani).
- **Undo:** hasil UNDO masuk lewat `enqueue_front`, sehingga langsung dilayani lagi (B dikembalikan di langkah 8, dilayani lagi di langkah 9).

## Tugas 6 — Kompleksitas Waktu

| Struktur | Operasi | Waktu | Alasan |
|---|---|---|---|
| LinkedQueue | `enqueue` | O(1) | Langsung menyambung di `rear.next` |
| LinkedQueue | `enqueue_front` | O(1) | Node baru diarahkan ke `front`, lalu `front` diganti |
| LinkedQueue | `dequeue` | O(1) | Cukup memajukan `front` |
| LinkedQueue | `peek` | O(1) | Langsung membaca `front.data` |
| Stack | `push` | O(1) | `append` di ujung list (amortized) |
| Stack | `pop` | O(1) | `list.pop()` di ujung list, tanpa menggeser elemen |
| Stack | `peek` | O(1) | Membaca `_items[-1]` |

(`display` pada keduanya O(n) karena menampilkan semua elemen.)

**Peran `front` dan `rear`:** singly linked list hanya bisa ditelusuri dari depan. Tanpa `rear`, `enqueue` harus menelusuri sampai node terakhir (O(n)). Dengan `rear` dan `front`, ujung belakang dan depan bisa diakses langsung, sehingga operasi utama Queue O(1).

## Tugas 7 — Linked List vs `list.pop(0)`
- `list.pop(0)` menggeser semua elemen sisa ke kiri, sehingga **O(n)** untuk setiap dequeue. Untuk n kali dequeue totalnya **O(n²)**.
- `dequeue` pada linked queue hanya memindahkan `front`, jadi **O(1)**. Total n kali dequeue hanya **O(n)**.
- Hal serupa berlaku untuk fitur Undo: `insert(0, x)` pada list O(n), sedangkan `enqueue_front` O(1).
- Kekurangan linked list hanya memori tambahan untuk referensi `next` di tiap node.

## Logika Program Soal 2
1. `LinkedQueue` mengelola node lewat `front`, `rear`, `count`: `enqueue` menyambung di rear, `dequeue` melepas dari front, `enqueue_front` menyisipkan di depan front.
2. `Stack` memakai list: `push` menambah di ujung dan `pop` mengambil dari ujung (LIFO).
3. LAYANI: ID hasil `dequeue` di-`push` ke Stack. UNDO: ID hasil `pop` dimasukkan lagi dengan `enqueue_front`.
4. Jika kosong, method menampilkan pesan dan mengembalikan `None`; LAYANI dan UNDO memeriksa `None` itu sehingga tidak error.
