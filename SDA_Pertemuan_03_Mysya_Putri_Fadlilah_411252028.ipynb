### 1. Membuat class Node dengan atribut data dan next

````python
from __future__ import annotations


class Node:
    def __init__(self, data: int) -> None:
        self.data: int = data
        self.next: Node | None = None
````

### 2. Membuat tiga node berisi 100, 200, dan 300

````python
node1 = Node(100)
node2 = Node(200)
node3 = Node(300)
````

### 3. Menghubungkan node pertama ke kedua, lalu node kedua ke ketiga

````python
node1.next = node2
node2.next = node3
````

### 4. Menampilkan data ketiga node dengan menelusuri dari node pertama (tanpa list Python)

````python
current: Node | None = node1
while current is not None:
    print(current.data)
    current = current.next
````

```

```

### 6. Mengubah next node pertama agar langsung menunjuk ke node ketiga

````python
node1.next = node3

current: Node | None = node1
while current is not None:
    print(current.data)
    current = current.next
````




Node yang dilewati adalah **node kedua (data 200)**, karena `next` dari node pertama sudah langsung menunjuk ke node ketiga.

### 7. Arti sebuah referensi `next`

`next` adalah referensi, yaitu penunjuk ke object Node berikutnya. `next` tidak menyimpan salinan node, hanya menunjuk ke object yang sama di memori, jadi node-node yang terpisah bisa tersambung menjadi satu rangkaian. Kalau `next` diganti, urutan rangkaiannya ikut berubah, seperti saat `node1.next` dipindah ke `node3` sehingga `node2` terlewati. Node terakhir punya `next = None` sebagai penanda akhir rangkaian.

# Praktik Mandiri 3.2 dan 3.3
Class `Node` dan `SinglyLinkedList` di bawah ini dipakai untuk Praktik Mandiri 3.2 dan 3.3.

````python
from __future__ import annotations


class Node:
    def __init__(self, data: int) -> None:
        self.data: int = data
        self.next: Node | None = None


class SinglyLinkedList:
    def __init__(self) -> None:
        self.head: Node | None = None
        self.size: int = 0

    def insert_beginning(self, data: int) -> None:
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node
        self.size += 1

    def append(self, data: int) -> None:
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
        else:
            current = self.head
            while current.next is not None:
                current = current.next
            current.next = new_node
        self.size += 1

    def insert_after(self, target: int, data: int) -> bool:
        current = self.head
        while current is not None:
            if current.data == target:
                new_node = Node(data)
                new_node.next = current.next
                current.next = new_node
                self.size += 1
                return True
            current = current.next
        return False

    def search(self, data: int) -> int:
        current = self.head
        position = 0
        while current is not None:
            if current.data == data:
                return position
            current = current.next
            position += 1
        return -1

    def delete(self, data: int) -> bool:
        if self.head is None:
            return False

        if self.head.data == data:
            self.head = self.head.next
            self.size -= 1
            return True

        previous = self.head
        current = self.head.next
        while current is not None:
            if current.data == data:
                previous.next = current.next
                self.size -= 1
                return True
            previous = current
            current = current.next
        return False

    def display(self) -> None:
        parts = []
        current = self.head
        while current is not None:
            parts.append(str(current.data))
            current = current.next
        parts.append("None")
        print(" -> ".join(parts))


def tampilkan(ll: SinglyLinkedList) -> None:
    ll.display()
    print("Jumlah node:", ll.size)
````

# Praktik Mandiri 3.2
## Praktik Mandiri 3.2 — Operasi Insert dan Search

### 1. Membuat SinglyLinkedList kosong

````python
ll = SinglyLinkedList()
````

### 2. Menambahkan data 15, 25, 35, 45 menggunakan append()

````python
ll.append(15)
ll.append(25)
ll.append(35)
ll.append(45)
ll.display()
````

### 3. Menambahkan 5 menggunakan insert_beginning()

### 4. Menyisipkan 30 setelah nilai 25

### 5. Menampilkan linked list

````python
tampilkan(ll)
````

### 6. Mencari nilai 30 dan 99

````python
print("Cari 30:", ll.search(30))
print("Cari 99:", ll.search(99))
````

**Hasil pencarian** (indeks dimulai dari 0, `-1` berarti tidak ditemukan):

- Nilai **30** ditemukan di indeks **3**.
- Nilai **99** **tidak ditemukan** (hasil `-1`), karena seluruh node sudah diperiksa dan tidak ada yang bernilai 99.

### 7. Dry run pencarian nilai 45

Isi list saat ini: `5 -> 15 -> 25 -> 30 -> 35 -> 45 -> None`

| Langkah | position | current.data | current.data == 45? |
|---|---|---|---|
| 1 | 0 | 5 | Tidak |
| 2 | 1 | 15 | Tidak |
| 3 | 2 | 25 | Tidak |
| 4 | 3 | 30 | Tidak |
| 5 | 4 | 35 | Tidak |
| 6 | 5 | 45 | **Ya, ditemukan** |

Hasil: `search(45)` mengembalikan **5** setelah **6 kali** pemeriksaan.

````python
target = 45
current = ll.head
position = 0

print(f"{'position':<10}{'current.data':<14}{'== 45?'}")
while current is not None:
    cocok = current.data == target
    print(f"{position:<10}{current.data:<14}{'Ya' if cocok else 'Tidak'}")
    if cocok:
        break
    current = current.next
    position += 1
````

### 8. Mengapa search memiliki kompleksitas O(n)

Linked list tidak bisa langsung loncat ke posisi tertentu seperti array. Pencarian harus mulai dari head, lalu berpindah satu per satu lewat `next` sambil membandingkan data. Pada kasus terburuk, nilai yang dicari ada di node terakhir atau tidak ada sama sekali, sehingga seluruh `n` node harus diperiksa. Jumlah langkahnya naik sebanding dengan jumlah node, maka kompleksitasnya **O(n)**.

## Praktik Mandiri 3.3 — Delete dan Edge Case

### 1. Menguji delete pada list kosong

````python
kosong = SinglyLinkedList()
print("Hasil delete(10):", kosong.delete(10))
tampilkan(kosong)
````

### 2. List dengan satu node, lalu hapus node tersebut

````python
satu = SinglyLinkedList()
satu.append(10)
print("Sebelum delete:")
tampilkan(satu)

print("\nHasil delete(10):", satu.delete(10))
print("Sesudah delete:")
tampilkan(satu)
print("head:", satu.head)
````

### 3. List minimal lima node: hapus head, node tengah, lalu node terakhir

````python
lima = SinglyLinkedList()
for nilai in [10, 20, 30, 40, 50]:
    lima.append(nilai)
print("Awal:")
tampilkan(lima)

print("\nHapus head (10):", lima.delete(10))
tampilkan(lima)

print("\nHapus node tengah (30):", lima.delete(30))
tampilkan(lima)

print("\nHapus node terakhir (50):", lima.delete(50))
tampilkan(lima)
````

### 4. Menghapus nilai yang tidak ada

````python
print("Hapus nilai 99:", lima.delete(99))
tampilkan(lima)
````

### 5. Isi list dan jumlah node setelah setiap operasi

Pada setiap pengujian di atas, isi list dan jumlah node sudah ditampilkan menggunakan fungsi `tampilkan()`. Ringkasan hasilnya:

| Operasi | Hasil delete | Isi list | Jumlah node |
|---|---|---|---|
| delete(10) pada list kosong | False | `None` | 0 |
| delete(10) pada list satu node | True | `None` | 0 |
| delete head (10) | True | `20 -> 30 -> 40 -> 50 -> None` | 4 |
| delete node tengah (30) | True | `20 -> 40 -> 50 -> None` | 3 |
| delete node terakhir (50) | True | `20 -> 40 -> None` | 2 |
| delete(99), tidak ada | False | `20 -> 40 -> None` | 2 |

### 6. Peran `previous` saat menghapus node yang bukan head

Linked list singly hanya punya penunjuk `next` ke depan, sehingga dari node yang akan dihapus kita tidak bisa kembali ke node sebelumnya. Padahal untuk menghapus sebuah node, `next` milik node sebelumnya harus diarahkan ke node sesudahnya (`previous.next = current.next`). Karena itu, saat menelusuri list kita menyimpan `previous` yang selalu tertinggal satu langkah di belakang `current`. Untuk head, tidak ada `previous`, jadi cukup memindahkan `head` ke `head.next`.

### 7. Pengujian tambahan untuk mencari kemungkinan kesalahan implementasi

**Tes A: nilai duplikat.** `delete()` hanya boleh menghapus kemunculan pertama.

````python
dup = SinglyLinkedList()
for nilai in [10, 20, 10]:
    dup.append(nilai)
print("Awal:")
tampilkan(dup)

print("\nHapus 10 (pertama):", dup.delete(10))
tampilkan(dup)

print("\nHapus 10 (kedua):", dup.delete(10))
tampilkan(dup)
````

**Tes B: list dua node, hapus node terakhir, lalu tambah data baru.**

````python
dua = SinglyLinkedList()
dua.append(1)
dua.append(2)
print("Hapus 2:", dua.delete(2))
tampilkan(dua)

dua.append(3)
print("\nSetelah append(3):")
tampilkan(dua)
````

**Tes C: search pada list kosong dan insert_after dengan target yang tidak ada.**

````python
uji = SinglyLinkedList()
print("search(5) pada list kosong:", uji.search(5))

uji.append(7)
print("insert_after(99, 8):", uji.insert_after(99, 8))
tampilkan(uji)
````
